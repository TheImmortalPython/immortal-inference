# Immortal Inference Quickstart: Pāṇinian Kāraka Logit Masking
This notebook demonstrates zero-drift linguistic containment and $O(1)$ inference constraint enforcement using `immortal-inference`.

```python
import torch
from pklm.constraints import LatticeConstraints, PaninianKarakaLogitsProcessor

# 1. Initialize Lattice Constraints
constraints = LatticeConstraints(
    kartā="AuthorizedUser",
    adhikaraṇa="SecureLocalEnvironment"
)

# 2. Define Sample Vocabulary Mapping
tokenizer_vocab = {
    "AuthorizedUser": 101,
    "SecureLocalEnvironment": 102,
    "standard_safe_output": 201,
    "unauthorized_drift_token": 301
}

processor = PaninianKarakaLogitsProcessor(
    constraints=constraints,
    tokenizer_vocab=tokenizer_vocab
)

# 3. Execute Logit Masking Filter
input_ids = torch.tensor([[101, 102]], dtype=torch.long)
logits = torch.tensor([[5.0, 5.0, 5.0, 5.0]], dtype=torch.float)

filtered_logits = processor(input_ids, logits)
print("Filtered Logits Output Vector:", filtered_logits)
```
